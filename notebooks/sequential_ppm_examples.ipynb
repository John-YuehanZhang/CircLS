# CircLS sequential-PPM examples


```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))

from circls.core.sequential_ppm_ls import PPMStep, SequentialPPMExperiment
from circls.core.routed_multi_patch_ls import PatchSpec, origin_of

D = 3
def spec(nm, a, b, o):
    return PatchSpec(nm, origin_of(a, b, D, seam=True), D, o)
```

## Example

```python
px4 = [spec("q1",   0, 0, "X_vertical"),
      spec("q3",   2, 1, "X_vertical"), 
      spec("q5",   0, 2, "X_vertical")]

seq4 = [PPMStep([("q1", "X"), ("q3", "X"), ("q5", "X")])]                                     

init4 = {f"q{i}": "Z" for i in range(1, 6)}
exp4 = SequentialPPMExperiment(
    px4, seq4, initial_states=init4, final_measure_states=init4,
    rounds=D, rounds_init=2, auto_rotate=True, rotation_kind="litinski", rotate_saving_threshold=1)
c4 = exp4.build()
print("rotation log      :", exp4.rotation_log)
print("per-step schedule :", dict(exp4._sched))
print("per-step corridor :", {i: sorted(r.tree) for i, r in enumerate(exp4._routes)})
print("num_ticks:", c4.num_ticks, " observables:", c4.num_observables)
```
